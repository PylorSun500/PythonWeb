## 数据库框架 `PyMySQL`
很重要的框架。连接了 Python 和 MySQL 。实际上不止 Flask 用得上。

### 基本准备

导入包 `pymysql`

### 连接：`pymysql.connect`

```py
conn = pymysql.connect(host="", user="", passwd="", db="", port=, charset="", cursorclass="", )
```

顺序如上。

### 获取 cursor ：`cursor()`

```py
cursor = conn.cursor
```

### 执行 SQL 语句

1. `cursor.execute(operation[,params])` 执行一条 SQL 语句

    - `oparation` : SQL 语句
    - `params` : 参数
    
    返回整数，表示语句影响的行数

2. `cursor.executemany(operation[,seq_of_params])` 执行多条 SQL 语句

    - `oparation` : SQL 语句
    - `params` : 序列

    返回整数，表示语句影响的行数

3. `cursor.fetchall()` 从结果集返回所有数据

4. `cursor.fetchone()` 从结果集返回一条记录的序列。无则返回 `none` 。

5. `cursor.fetchmany([size=cursor.arraysize])` 从结果集返回 <= `size` 的记录数序列，无则返回空序列，size 默认为整个游标的行数。

6. `cursor.rowcount` 最近一次执行数据库查询命令后，返回影响的行数。

### 关闭 cursor

`cursor.close()`

### 关闭连接

`conn.close()`
